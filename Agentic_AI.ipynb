Import Libraries

```python
import os
from datetime import datetime

from google.colab import drive
from google.colab import userdata

from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter

from langchain_huggingface import HuggingFaceEmbeddings
from langchain_chroma import Chroma

from langchain_google_genai import ChatGoogleGenerativeAI

from langchain_core.prompts import ChatPromptTemplate
from langchain_core.output_parsers import StrOutputParser
from langchain_core.tools import tool

from langchain_classic.retrievers import MultiQueryRetriever

from langchain.agents import create_agent
from langgraph.checkpoint.memory import InMemorySaver
```

Mount Google Drive

```python
drive.mount('/content/drive')
```

```python
pdf_path="/content/drive/MyDrive/Badminton_Complete_Knowledge_Base_10plus_pages.pdf"
```

Load PDF

```python
loader = PyPDFLoader(pdf_path)

documents = loader.load()

print("Number of pages:", len(documents))
```

View PDF Content

```python
print(documents[0].page_content[:2000])
```

Split Documents

```python
text_splitter = RecursiveCharacterTextSplitter(
    chunk_size=1000,
    chunk_overlap=200
)

chunks = text_splitter.split_documents(documents)

print("Number of chunks:", len(chunks))
```

Create Embeddings

```python
embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)
```

Create Chroma Vector Database

```python
vectorstore = Chroma.from_documents(
    documents=chunks,
    embedding=embeddings,
    collection_name="badminton_knowledge"
)
```

Create Retriever

```python
retriever = vectorstore.as_retriever(
    search_type="similarity",
    search_kwargs={"k": 4}
)
```

Test Retriever

```python
query = "What is a smash in badminton?"

results = retriever.invoke(query)

for i, doc in enumerate(results):
    print("\nRESULT", i + 1)
    print(doc.page_content[:1000])
```

```python
GEMINI_API_KEY = userdata.get("Key")

print("API key loaded successfully")
```

Create Gemini LLM

```python
llm = ChatGoogleGenerativeAI(
    model="gemini-3.8-flash",
    temperature=0,
    google_api_key=GEMINI_API_KEY
)
```

MultiQueryRetriever

```python
multi_query_retriever = MultiQueryRetriever.from_llm(
    retriever=retriever,
    llm=llm
)
```

Test Gemini

```python
response = llm.invoke("What is badminton?")

print(response.content)
```

Test MultiQueryRetriever

```python
query = "How should a player perform a smash?"

results = multi_query_retriever.invoke(query)

for i, doc in enumerate(results):
    print("\nRESULT", i + 1)
    print(doc.page_content[:1000])
```

RAG Prompt

```python
prompt = ChatPromptTemplate.from_template("""
You are a badminton knowledge assistant.

Answer the user's question using only the information provided in the context.

If the answer is not available in the context, say:
"I could not find this information in the badminton knowledge base."

Do not make up information.

Context:
{context}

Question:
{question}

Answer:
""")
```

Format Documents

```python
def format_docs(docs):
    return "\n\n".join(doc.page_content for doc in docs)
```

RAG Function

```python
def rag(question):

    docs = multi_query_retriever.invoke(question)

    context = format_docs(docs)

    chain = prompt | llm | StrOutputParser()

    answer = chain.invoke({
        "context": context,
        "question": question
    })

    return answer
```

Test RAG

```python
question = "What is a smash in badminton?"

answer = rag(question)

print(answer)
```

PART 2: Convert RAG into Tool 1

Tool 1 - RAG Tool

```python
print(type(rag))
```

```python
from langchain_core.tools import tool

@tool
def rag_tool(query: str) -> str:
    """
    Answer badminton questions using the badminton knowledge base.
    """
    return rag(query)
```

```python
print(rag_tool.invoke("What is a smash in badminton?"))
```

PART 3: Tool 2 - Calculator

Calculator Tool

```python
@tool
def calculator(expression: str) -> str:
    """
    Calculate a mathematical expression.
    """
    try:
        result = eval(expression, {"__builtins__": {}}, {})
        return str(result)
    except:
        return "Invalid mathematical expression."
```

```python
print(calculator.invoke("125 * 48"))
```

Tool 3 - Current Date

Current Date Tool

```python
@tool
def current_date() -> str:
    """
    Return today's current date.
    """
    return datetime.now().strftime("%d-%m-%Y")
```

```python
print(current_date.invoke(""))
```

PART 5: Check Our 3 Tools

```python
tools = [
    rag_tool,
    calculator,
    current_date
]

print("Number of tools:", len(tools))

for tool in tools:
    print(tool.name)
```

PART 6: Create Agent Memory

Memory

```python
memory = InMemorySaver()
```

PART 7: Create Agent

```python
agent = create_agent(
    model=llm,
    tools=tools,
    checkpointer=memory,
    system_prompt="""
You are a helpful badminton knowledge assistant.

You have access to exactly three tools:

1. rag_tool - for badminton knowledge from the provided document.
2. calculator - for mathematical calculations.
3. current_date - for today's date.

Automatically select the appropriate tool based on the user's question.

Use rag_tool for badminton questions.
Use calculator for mathematical questions.
Use current_date for date questions.

Answer clearly and do not invent information.
"""
)
```

PART 8: Function to Ask the Agent

```python
def ask_agent(question, thread_id):

    response = agent.invoke(
        {
            "messages": [
                {
                    "role": "user",
                    "content": question
                }
            ]
        },
        config={
            "configurable": {
                "thread_id": thread_id
            }
        }
    )

    return response["messages"][-1].content
```
